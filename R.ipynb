````r
# Подготовка данных
experience_map <- c("Нет опыта" = 0,
                    "От 1 года до 3 лет" = 2,
                    "От 3 до 6 лет" = 4.5,
                    "Более 6 лет" = 7)
df_excel$Experience_years <- experience_map[df_excel$Experience]
df_excel$Experience_years[is.na(df_excel$Experience_years)] <- 0
df_excel$Is_remote <- as.integer(grepl("Удаленная", df_excel$Schedule))

# Модель
model <- lm(To_numeric ~ From_numeric + Experience_years + Is_remote, data = df_excel)
summary(model)
````

````r
extract_main_role <- function(roles_list) {
  # Если значение пустое
  if (is.na(roles_list)) {
    return("Другое")
  }

  # Преобразуем в символьную строку
  roles_str <- as.character(roles_list)

  # Убираем квадратные скобки и кавычки
  roles_str <- gsub("\\[|\\]", "", roles_str)
  roles_str <- gsub("'", "", roles_str)
  roles_str <- gsub('"', "", roles_str)

  # Разбиваем по запятой
  roles <- trimws(unlist(strsplit(roles_str, ",")))

  if (length(roles) == 0) {
    return("Другое")
  }

  main_role <- roles[1]

  # Упрощаем названия
  if (grepl("Программист|разработчик", main_role, ignore.case = TRUE)) {
    return("Разработчик")
  } else if (grepl("Аналитик", main_role)) {
    return("Аналитик")
  } else if (grepl("Тестировщик|QA", main_role)) {
    return("Тестировщик")
  } else if (grepl("Дизайнер|художник", main_role, ignore.case = TRUE)) {
    return("Дизайнер")
  } else if (grepl("Менеджер|Руководитель", main_role)) {
    return("Менеджер")
  } else if (grepl("Администратор|DevOps|Системный", main_role)) {
    return("Администратор/DevOps")
  } else if (grepl("Специалист|Инженер", main_role)) {
    return("Специалист")
  } else {
    return(main_role)
  }
}

# Применяем функцию
df_excel$MainRole <- sapply(df_excel$Professional_roles, extract_main_role)

# Смотрим новое распределение
print("Распределение вакансий по основным ролям:")
table(df_excel$MainRole)

print("Структура MainRole:")
str(df_excel$MainRole)
````

````r
# Создаем среднюю зарплату
df_excel$SalaryMid <- (df_excel$From_numeric + df_excel$To_numeric) / 2
df_excel$SalaryMid[is.na(df_excel$SalaryMid)] <- df_excel$From_numeric[is.na(df_excel$SalaryMid)]
df_excel$SalaryMid[is.na(df_excel$SalaryMid)] <- df_excel$To_numeric[is.na(df_excel$SalaryMid)]
df_excel$SalaryMid[is.na(df_excel$SalaryMid)] <- median(df_excel$SalaryMid, na.rm = TRUE)

# Кодируем опыт работы
exp_map <- c("Нет опыта" = 0,
             "От 1 года до 3 лет" = 1,
             "От 3 до 6 лет" = 2,
             "Более 6 лет" = 3)
df_excel$ExperienceCode <- exp_map[df_excel$Experience]
df_excel$ExperienceCode[is.na(df_excel$ExperienceCode)] <- 0

# Кодируем график работы
schedule_map <- c("Удаленная работа" = 0,
                  "Полный день" = 1,
                  "Сменный график" = 2,
                  "Гибкий график" = 3)
df_excel$ScheduleCode <- schedule_map[df_excel$Schedule]
df_excel$ScheduleCode[is.na(df_excel$ScheduleCode)] <- 1

# Кодируем основную роль
if (is.list(df_excel$MainRole)) {
  df_excel$MainRole <- unlist(df_excel$MainRole)
}

df_excel$RoleCode <- as.numeric(as.factor(df_excel$MainRole)) - 1

# Количество навыков
df_excel$SkillsCount <- sapply(df_excel$Keys, function(x) {
  if (is.na(x)) return(0)
  skills_str <- gsub("'", '"', x)
  skills <- tryCatch(eval(parse(text = skills_str)), error = function(e) c())
  return(length(skills))
})

# Отбор признаков
features <- df_excel[, c("SalaryMid", "ExperienceCode", "ScheduleCode", "SkillsCount", "RoleCode")]

# Масштабирование (обязательно для K-means)
features_scaled <- scale(features)

# Проверка
print("Размер данных:")
dim(features_scaled)
print("Признаки:")
colnames(features_scaled)
````

````r
# Переименовываем столбцы
colnames(df_excel)[1] <- "Ids"
colnames(df_excel)[2] <- "Employer"
colnames(df_excel)[3] <- "Name"
colnames(df_excel)[4] <- "Salary"
colnames(df_excel)[5] <- "From"
colnames(df_excel)[6] <- "To"
colnames(df_excel)[7] <- "Experience"
colnames(df_excel)[8] <- "Schedule"
colnames(df_excel)[9] <- "Keys"
colnames(df_excel)[10] <- "Description"
colnames(df_excel)[11] <- "Area"
colnames(df_excel)[12] <- "Professional_roles"
colnames(df_excel)[13] <- "Specializations"
colnames(df_excel)[14] <- "Profarea_names"
colnames(df_excel)[15] <- "Published_at"

# Преобразуем From и To в числовые
df_excel$From_numeric <- as.numeric(df_excel$From)
df_excel$To_numeric <- as.numeric(df_excel$To)

# Извлекаем основную роль (простой способ)
df_excel$MainRole <- "Другое"
df_excel$MainRole[grepl("Программист|разработчик", df_excel$Professional_roles, ignore.case = TRUE)] <- "Разработчик"
df_excel$MainRole[grepl("Аналитик", df_excel$Professional_roles)] <- "Аналитик"
df_excel$MainRole[grepl("Тестировщик|QA", df_excel$Professional_roles)] <- "Тестировщик"
df_excel$MainRole[grepl("Дизайнер|художник", df_excel$Professional_roles, ignore.case = TRUE)] <- "Дизайнер"
df_excel$MainRole[grepl("Менеджер|Руководитель", df_excel$Professional_roles)] <- "Менеджер"
df_excel$MainRole[grepl("Администратор|DevOps|Системный", df_excel$Professional_roles)] <- "Администратор/DevOps"
df_excel$MainRole[grepl("Специалист|Инженер", df_excel$Professional_roles)] <- "Специалист"

# Создаем среднюю зарплату
df_excel$SalaryMid <- (df_excel$From_numeric + df_excel$To_numeric) / 2
df_excel$SalaryMid[is.na(df_excel$SalaryMid)] <- df_excel$From_numeric[is.na(df_excel$SalaryMid)]
df_excel$SalaryMid[is.na(df_excel$SalaryMid)] <- df_excel$To_numeric[is.na(df_excel$SalaryMid)]
df_excel$SalaryMid[is.na(df_excel$SalaryMid)] <- median(df_excel$SalaryMid, na.rm = TRUE)

# Кодируем опыт работы
exp_map <- c("Нет опыта" = 0,
             "От 1 года до 3 лет" = 1,
             "От 3 до 6 лет" = 2,
             "Более 6 лет" = 3)
df_excel$ExperienceCode <- exp_map[df_excel$Experience]
df_excel$ExperienceCode[is.na(df_excel$ExperienceCode)] <- 0

# Кодируем график работы
schedule_map <- c("Удаленная работа" = 0,
                  "Полный день" = 1,
                  "Сменный график" = 2,
                  "Гибкий график" = 3)
df_excel$ScheduleCode <- schedule_map[df_excel$Schedule]
df_excel$ScheduleCode[is.na(df_excel$ScheduleCode)] <- 1

# Кодируем основную роль
role_map <- setNames(0:(length(unique(df_excel$MainRole)) - 1), unique(df_excel$MainRole))
df_excel$RoleCode <- role_map[df_excel$MainRole]

# Количество навыков (упрощенный способ)
df_excel$SkillsCount <- sapply(df_excel$Keys, function(x) {
  if (is.na(x)) return(0)
  # Считаем количество элементов через запятые
  x_str <- as.character(x)
  count <- nchar(gsub("[^,]", "", x_str)) + 1
  return(count)
})
````

````r
# 1. Очищаем данные

# Преобразуем в data.frame если это data.table
if(is.data.table(features)) {
  features <- as.data.frame(features)
}

print(paste("Исходное количество строк:", nrow(features)))

# Проверяем каждый столбец на константность
print("\nПроверка столбцов на константность:")
constant_cols <- c()
for(col in names(features)) {
  if(length(unique(features[[col]])) == 1) {
    constant_cols <- c(constant_cols, col)
    print(paste("Столбец", col, "имеет все одинаковые значения:", unique(features[[col]])[1]))
  }
}

# Удаляем константные столбцы
if(length(constant_cols) > 0) {
  print(paste("\nУдаляем константные столбцы:", paste(constant_cols, collapse = ", ")))
  features <- features[, !names(features) %in% constant_cols, drop = FALSE]
} else {
  print("Константных столбцов не найдено")
}

# Находим и удаляем строки с проблемами
valid_rows <- complete.cases(features)

# Проверяем на Inf и NaN
for(col in names(features)) {
  valid_rows <- valid_rows & !is.infinite(features[[col]]) & !is.nan(features[[col]])
}

print(paste("\nУдалено строк с проблемами:", sum(!valid_rows)))
print(paste("Осталось строк:", sum(valid_rows)))

if(sum(valid_rows) == 0) {
  stop("Нет валидных строк для анализа!")
}

# Оставляем только чистые строки
features_clean <- features[valid_rows, ]

# Очищаем df_excel
if(exists("df_excel")) {
  df_clean <- df_excel[valid_rows, ]
} else if(exists("df")) {
  df_clean <- df[valid_rows, ]
}

# 2. Масштабируем чистые данные
print("\nМасштабирование данных...")
features_scaled <- scale(features_clean)

# Проверяем результат
print("Проверка масштабированных данных:")
print(paste("NA:", any(is.na(features_scaled))))
print(paste("Inf:", any(is.infinite(features_scaled))))
print(paste("NaN:", any(is.nan(features_scaled))))

# Если есть NaN, удаляем соответствующие столбцы
if(any(is.nan(features_scaled))) {
  print("\nОбнаружены NaN в масштабированных данных. Удаляем проблемные столбцы...")

  # Находим столбцы с NaN
  cols_with_nan <- apply(features_scaled, 2, function(x) any(is.nan(x)))
  print(paste("Столбцы с NaN:", paste(names(which(cols_with_nan)), collapse = ", ")))

  # Удаляем эти столбцы
  features_clean <- features_clean[, !cols_with_nan, drop = FALSE]

  # Повторно масштабируем
  features_scaled <- scale(features_clean)

  print("Повторная проверка:")
  print(paste("NA:", any(is.na(features_scaled))))
  print(paste("NaN:", any(is.nan(features_scaled))))
}

# 3. K-means кластеризация
if(!any(is.na(features_scaled)) && !any(is.nan(features_scaled))) {
  set.seed(42)
  kmeans_result <- kmeans(features_scaled, centers = 4, nstart = 25)

  # Добавляем кластеры
  df_clean$Cluster_Kmeans <- kmeans_result$cluster

  # Результаты
  print("Распределение вакансий по кластерам (K-means):")
  print(table(df_clean$Cluster_Kmeans))

  print("\nДоля каждого кластера:")
  print(prop.table(table(df_clean$Cluster_Kmeans)))

  print("\nРаспределение профессиональных ролей по кластерам:")
  cluster_roles <- table(df_clean$Cluster_Kmeans, df_clean$MainRole)
  print(cluster_roles)

  # Визуализация
  barplot(table(df_clean$Cluster_Kmeans),
          main = "Распределение по кластерам",
          xlab = "Кластер",
          ylab = "Количество вакансий",
          col = "steelblue")

} else {
  print("ОШИБКА: Данные всё ещё содержат NA/NaN после удаления константных столбцов")
  print("Структура features_scaled:")
  print(str(features_scaled))
}
````

````r
# Анализ центроидов
centroids <- kmeans_result$centers

# Проверяем, какие столбцы есть в центроидах
print("Доступные столбцы в центроидах:")
print(colnames(centroids))

# Функция для обратного масштабирования
reverse_scale <- function(scaled_values, original_col, features_scaled_obj) {
  # Проверяем, существует ли столбец в атрибутах масштабирования
  if(original_col %in% names(attr(features_scaled_obj, "scaled:center"))) {
    means <- attr(features_scaled_obj, "scaled:center")[original_col]
    sds <- attr(features_scaled_obj, "scaled:scale")[original_col]
    return(scaled_values * sds + means)
  } else {
    return(rep(NA, length(scaled_values)))
  }
}

# Создаем таблицу характеристик кластеров динамически
cluster_summary <- data.frame(Cluster = 1:nrow(centroids))

# Добавляем каждый столбец, если он существует
if("SalaryMid" %in% colnames(centroids)) {
  cluster_summary$SalaryMid_k <- round(
    reverse_scale(centroids[, "SalaryMid"], "SalaryMid", features_scaled) / 1000, 0
  )
}

if("ExperienceCode" %in% colnames(centroids)) {
  cluster_summary$ExperienceCode <- round(
    reverse_scale(centroids[, "ExperienceCode"], "ExperienceCode", features_scaled), 1
  )
}

if("ScheduleCode" %in% colnames(centroids)) {
  cluster_summary$ScheduleCode <- round(
    reverse_scale(centroids[, "ScheduleCode"], "ScheduleCode", features_scaled), 1
  )
}

if("SkillsCount" %in% colnames(centroids)) {
  cluster_summary$SkillsCount <- round(
    reverse_scale(centroids[, "SkillsCount"], "SkillsCount", features_scaled), 1
  )
}

if("RoleCode" %in% colnames(centroids)) {
  cluster_summary$RoleCode <- round(
    reverse_scale(centroids[, "RoleCode"], "RoleCode", features_scaled), 1
  )
}

print("Характеристики кластеров:")
print(cluster_summary)
````

````r
# Очищаем окружение (опционально)
# rm(list = ls())

# Загружаем библиотеки
library(readxl)
library(caret)

df_excel <- read_excel("/content/IT_vacancies_full.xlsx")

# Проверяем загрузку
print(paste("Загружено строк:", nrow(df_excel)))
print(paste("Загружено столбцов:", ncol(df_excel)))
print("Имена столбцов:")
print(names(df_excel))

# Теперь преобразуем в data.frame
df_excel <- as.data.frame(df_excel)

print("Структура данных:")
str(df_excel)

# Теперь продолжаем подготовку данных
# Преобразование Experience
df_excel$Experience <- as.factor(df_excel$Experience)

# преобразование в numeric
df_excel$From <- as.numeric(as.character(df_excel$From))
df_excel$To <- as.numeric(as.character(df_excel$To))

# Проверяем, сколько NA появилось после преобразования
print(paste("NA в From:", sum(is.na(df_excel$From))))
print(paste("NA в To:", sum(is.na(df_excel$To))))
print(paste("NA в Experience:", sum(is.na(df_excel$Experience))))

# Удаляем строки с NA ТОЛЬКО в критических столбцах
df_excel_clean <- df_excel[complete.cases(df_excel[, c("From", "To", "Experience")]), ]

print(paste("Размер после очистки:", nrow(df_excel_clean), "строк"))

# Проверяем, остались ли данные
if(nrow(df_excel_clean) == 0) {
  print("ВНИМАНИЕ: Все строки содержат NA в критических столбцах!")
  print("Статистика по From:")
  print(summary(df_excel$From))
  print("Статистика по To:")
  print(summary(df_excel$To))
  print("Статистика по Experience:")
  print(summary(df_excel$Experience))
} else {
  # Создание целевой переменной
  df_excel_clean$Salary_avg <- (df_excel_clean$From + df_excel_clean$To) / 2
  df_excel_clean$Target <- as.factor(ifelse(df_excel_clean$Salary_avg > 130000, 1, 0))

  cat("\nРаспределение целевой переменной:\n")
  print(table(df_excel_clean$Target))

  # Кодируем Experience в числовой формат
  df_excel_clean$Experience_num <- as.numeric(df_excel_clean$Experience)

  # Выбираем признаки
  X <- as.matrix(df_excel_clean[, c("From", "To", "Experience_num")])
  y <- df_excel_clean$Target

  # Масштабирование
  scaler <- preProcess(X, method = c("center", "scale"))
  X_scaled <- predict(scaler, X)

  print("Масштабирование выполнено успешно!")
  print(paste("Размер X_scaled:", dim(X_scaled)))
  print("Первые несколько строк X_scaled:")
  print(head(X_scaled))
}
````

````r
# Загружаем необходимые библиотеки
library(caret)
library(ggplot2)
library(pROC)
library(MASS)  # для lda()

# Убедимся, что данные готовы
if(!exists("X_scaled")) {
  stop("Сначала нужно создать X_scaled! Выполните предыдущие шаги подготовки данных.")
}

if(!exists("df_excel_clean")) {
  stop("Нет очищенных данных!")
}

# Разделение данных на обучающую и тестовую выборки
set.seed(123)
train_index <- createDataPartition(df_excel_clean$Target, p = 0.7, list = FALSE)
X_train <- X_scaled[train_index, ]
X_test <- X_scaled[-train_index, ]
y_train <- df_excel_clean$Target[train_index]
y_test <- df_excel_clean$Target[-train_index]

print(paste("Обучающая выборка:", nrow(X_train), "строк"))
print(paste("Тестовая выборка:", nrow(X_test), "строк"))

# Обучение модели LDA
lda_model <- lda(x = X_train, grouping = y_train)

# Предсказания на тестовой выборке
y_pred <- predict(lda_model, X_test)

# Матрица ошибок (Confusion Matrix)
cm <- confusionMatrix(y_pred$class, y_test)

# Вывод матрицы ошибок
print("Матрица ошибок (Confusion Matrix):")
print(cm$table)

# Метрики качества
print("\nМетрики классификации:")
print(cm$overall)

print("\nМетрики по классам:")
print(cm$byClass)

# Визуализация матрицы ошибок
cm_matrix <- cm$table
cm_df <- as.data.frame(as.table(cm_matrix))
colnames(cm_df) <- c("Actual", "Predicted", "Freq")

# Тепловая карта матрицы ошибок
ggplot(cm_df, aes(x = Predicted, y = Actual, fill = Freq)) +
  geom_tile() +
  geom_text(aes(label = Freq), size = 8) +
  scale_fill_gradient(low = "white", high = "steelblue") +
  labs(title = "Confusion Matrix - LDA",
       x = "Predicted Class",
       y = "Actual Class") +
  theme_minimal() +
  theme(plot.title = element_text(hjust = 0.5, size = 16))

# Сохраняем график
ggsave("confusion_matrix_lda_r.png", width = 8, height = 6)

# ROC-кривая
# Для LDA нам нужны вероятности принадлежности к классу "1"
y_pred_proba <- predict(lda_model, X_test)$posterior[, "1"]

# Строим ROC-кривую
roc_curve <- roc(y_test, y_pred_proba)
auc <- auc(roc_curve)

# Визуализация ROC-кривой
plot(roc_curve,
     main = paste("ROC Curve - LDA Classifier (AUC =", round(auc, 3), ")"),
     col = "blue",
     lwd = 2)

# Добавляем AUC на график
text(0.6, 0.2, paste("AUC =", round(auc, 3)), cex = 1.2)

# Сохраняем ROC-кривую
png("roc_curve_lda_r.png", width = 800, height = 600)
plot(roc_curve,
     main = paste("ROC Curve - LDA Classifier (AUC =", round(auc, 3), ")"),
     col = "blue",
     lwd = 2)
text(0.6, 0.2, paste("AUC =", round(auc, 3)), cex = 1.2)
dev.off()

print(paste("AUC =", round(auc, 3)))
````

````r
# График разделения классов в пространстве LDA
lda_projection <- as.data.frame(predict(lda_model, X_train)$x)
lda_projection$Target <- y_train

ggplot(lda_projection, aes(x = LD1, fill = Target)) +
  geom_histogram(alpha = 0.6, position = "identity", bins = 30) +
  labs(title = "LDA Projection - Separation of Classes",
       x = "First Linear Discriminant",
       y = "Count") +
  theme_minimal() +
  theme(plot.title = element_text(hjust = 0.5))

ggsave("lda_projection_r.png", width = 10, height = 6)

# Получаем коэффициенты линейных дискриминантов
lda_coef <- data.frame(
  Feature = colnames(X_train),
  LD1 = lda_model$scaling[, 1]
)

print("Коэффициенты линейных дискриминантов:")
print(lda_coef)

# Визуализация важности признаков
ggplot(lda_coef, aes(x = reorder(Feature, abs(LD1)), y = LD1, fill = LD1 > 0)) +
  geom_bar(stat = "identity") +
  coord_flip() +
  labs(title = "Feature Importance in LDA",
       x = "Features",
       y = "LD1 Coefficient") +
  theme_minimal() +
  theme(legend.position = "none")

ggsave("lda_feature_importance.png", width = 8, height = 6)

# Метрики качества модели
accuracy <- cm$overall["Accuracy"]
precision <- cm$byClass["Precision"]
recall <- cm$byClass["Sensitivity"]
specificity <- cm$byClass["Specificity"]
f1_score <- cm$byClass["F1"]

# ИНТЕРПРЕТАЦИЯ
cat("\n")
cat("1. ВАЖНОСТЬ ПРИЗНАКОВ:\n")

if (abs(lda_coef[1,2]) > abs(lda_coef[2,2])) {
  cat("   - From (нижняя граница зарплаты) имеет большее влияние на классификацию\n")
  cat(paste0("     Коэффициент: ", round(lda_coef[1,2], 3), "\n"))
} else {
  cat("   - To (верхняя граница зарплаты) имеет большее влияние на классификацию\n")
  cat(paste0("     Коэффициент: ", round(lda_coef[2,2], 3), "\n"))
}

if(nrow(lda_coef) > 2) {
  cat(paste0("   - ", lda_coef[3,1], " (опыт) имеет коэффициент: ", round(lda_coef[3,2], 3), "\n"))
}

cat("\n2. КАЧЕСТВО МОДЕЛИ:\n")
cat(paste0("   - Точность (Accuracy): ", round(accuracy*100, 1), "%\n"))
cat(paste0("   - Полнота (Recall/Sensitivity): ", round(recall*100, 1), "%\n"))
cat(paste0("   - Специфичность (Specificity): ", round(specificity*100, 1), "%\n"))
cat(paste0("   - Точность предсказания (Precision): ", round(precision*100, 1), "%\n"))
cat(paste0("   - F1-мера: ", round(f1_score, 3), "\n"))
cat(paste0("   - AUC = ", round(auc, 3), " - ",
           ifelse(auc > 0.9, "превосходное",
           ifelse(auc > 0.8, "отличное",
           ifelse(auc > 0.7, "хорошее",
           ifelse(auc > 0.6, "удовлетворительное", "плохое")))),
           " качество\n"))

cat("\n3. АНАЛИЗ ОШИБОК:\n")
cm_matrix <- cm$table
false_positives <- cm_matrix[2,1]
false_negatives <- cm_matrix[1,2]
total <- sum(cm_matrix)

cat(paste0("   - Ложноположительные ошибки (FP): ", false_positives,
           " (", round(false_positives/total*100, 1), "%)\n"))
cat(paste0("   - Ложноотрицательные ошибки (FN): ", false_negatives,
           " (", round(false_negatives/total*100, 1), "%)\n"))

if(false_positives > false_negatives) {
  cat("   - Модель чаще ошибается, предсказывая высокую зарплату там, где её нет\n")
} else {
  cat("   - Модель чаще пропускает случаи высокой зарплаты\n")
}

cat("\n4. РЕКОМЕНДАЦИИ:\n")
if (recall > precision) {
  cat("   - Модель лучше находит истинные случаи (высокий recall)\n")
  cat("   - Рекомендуется использовать для поиска кандидатов на высокие зарплаты\n")
} else {
  cat("   - Модель редко ошибается с предсказаниями (высокая precision)\n")
  cat("   - Рекомендуется для автоматического отбора кандидатов\n")
}

if (auc > 0.8) {
  cat("   - Модель хорошо разделяет классы, можно использовать в production\n")
} else if (auc > 0.7) {
  cat("   - Модель приемлемого качества, но требует доработки\n")
} else {
  cat("   - Качество модели низкое, требуется улучшение признаков\n")
}

cat("\n5. СТАТИСТИКА ПО КЛАССАМ:\n")
cat(paste0("   - Класс 0 (низкая зарплата): ", sum(y_train == 0), " наблюдений\n"))
cat(paste0("   - Класс 1 (высокая зарплата): ", sum(y_train == 1), " наблюдений\n"))
class_balance <- sum(y_train == 1) / length(y_train)
cat(paste0("   - Баланс классов: ", round(class_balance*100, 1), "% высокой зарплаты\n"))

if(class_balance < 0.3 || class_balance > 0.7) {
  cat("   - ВНИМАНИЕ: Данные несбалансированы! Рекомендуется использовать взвешивание классов\n")
}

# Сохранение модели и скейлера
saveRDS(lda_model, "lda_model_r.rds")
saveRDS(scaler, "scaler_r.rds")
cat("\n✓ Модель и скейлер сохранены в файлы:\n")
cat("   - lda_model_r.rds\n")
cat("   - scaler_r.rds\n")

# Сохраняем метрики в CSV
metrics_df <- data.frame(
  Metric = c("Accuracy", "Precision", "Recall", "Specificity", "F1", "AUC"),
  Value = c(accuracy, precision, recall, specificity, f1_score, auc)
)
write.csv(metrics_df, "lda_metrics.csv", row.names = FALSE)
cat("\n✓ Метрики сохранены в lda_metrics.csv\n")
````
